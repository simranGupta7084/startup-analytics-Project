Inspect null handling, categorical encoding, and numeric scaling.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT / 'src'))

import pandas as pd
from preprocess import load_dataset, split_features_target, build_preprocessor, save_preprocessor
```

```python
df = load_dataset()
bundle = split_features_target(df)
bundle.features.head()
```

```python
numeric_cols = bundle.features.select_dtypes(include=['number', 'bool']).columns.tolist()
categorical_cols = [col for col in bundle.features.columns if col not in numeric_cols]
print('Numeric:', numeric_cols)
print('Categorical:', categorical_cols)
```

```python
preprocessor = build_preprocessor(bundle.features)
transformed = preprocessor.fit_transform(bundle.features, bundle.target)
feature_names = preprocessor.get_feature_names_out()
processed = pd.DataFrame(transformed, columns=feature_names)
processed.head()
```

```python
processed.isna().sum().sum(), processed.shape
```

```python
save_preprocessor(preprocessor, bundle.features.columns)
print('Preprocessor saved.')
```
