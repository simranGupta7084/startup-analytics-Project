# 01 - Exploratory Data Analysis

EDA for the Kaggle E-Commerce Customer Churn dataset.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT / 'src'))

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from preprocess import DATA_PATH, load_dataset

sns.set_theme(style='whitegrid')
```

```python
df = load_dataset(DATA_PATH)
df.head()
```

```python
df.info()
df.describe(include='all').T
```

```python
missing = df.isna().mean().mul(100).sort_values(ascending=False)
missing[missing > 0]
```

```python
churn_rate = df['Churn'].mean()
print(f'Churn rate: {churn_rate:.2%}')
sns.countplot(data=df, x='Churn')
plt.title('Churn Class Balance')
plt.show()
```

```python
numeric_cols = df.select_dtypes(include='number').columns.drop('Churn', errors='ignore')
corr = df[numeric_cols.tolist() + ['Churn']].corr(numeric_only=True)['Churn'].sort_values(ascending=False)
corr
```

```python
top_numeric = corr.drop('Churn', errors='ignore').abs().sort_values(ascending=False).head(6).index.tolist()
df[top_numeric + ['Churn']].hist(figsize=(12, 8), bins=20)
plt.tight_layout()
plt.show()
```

```python
categorical_cols = df.select_dtypes(exclude='number').columns.tolist()
for column in categorical_cols:
    summary = df.groupby(column, dropna=False)['Churn'].mean().sort_values(ascending=False)
    display(summary.to_frame('churn_rate'))
```
